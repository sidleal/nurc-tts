```python
import pandas as pd

# Carrega o arquivo CSV
df = pd.read_csv('nurc_tts_SELECIONADOS.csv')

# Exibe as 5 primeiras linhas para verificar se deu certo
df.head()
```

```python
import re
alphabet = r"ABCDEFGHIJKLMNOPQRSTUVWXYZÇÃÀÁÂÊÉÍÓÔÕÚÛabcdefghijklmnopqrstuvwxyzçãàáâêéíóôõũúû1234567890%\-\n/\\ "

def replace_special_tokens_and_normalize(text):
    text = text.lower()

    map_words = {
        "éh": "eh",
        "ehm": "eh",
        "ehn": "eh",
        "hum": "uh",
        "hm": "uh",
        "uhm": "uh",
        "hã": "ah",
        "ãh": "ah",
        "ã":  "ah",
        "hmm": "uh",
        "mm": "uh",
        "mhm": "uh"
    }

    text = re.sub("h+", "h", text)
    text = re.sub("[^{}]".format(alphabet+" "), " ", text)
    text = re.sub("[ ]+", " ", text)

    words = text.split(' ')
    new_words = []
    for word in words:
        if word == '' or word == ' ':
            continue
        if word in map_words:
            new_words.append(map_words[word])
        else:
            new_words.append(word)

    return " ".join(new_words)
```

```python
# Aplica a função na coluna 'text' e salva o resultado nela mesma
df['text'] = df['text'].apply(replace_special_tokens_and_normalize)

df.head()
```

```python
# Salva o DataFrame em um novo arquivo CSV
df.to_csv('nurc_tts_SELECIONADOS_normalizado.csv', index=False, encoding='utf-8')
```
