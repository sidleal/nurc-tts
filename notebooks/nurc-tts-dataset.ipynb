```python
from datasets import load_dataset
import pandas as pd
```

```python
ds = load_dataset('sidleal/nurc_tts_24khz')
print(ds)
```

```python
splits = ['sao_paulo', 'recife']
all_metadata = []

for split_name in splits:
    ds = load_dataset('sidleal/nurc_tts_24khz', split=split_name)
    metadata = ds.remove_columns(['audio'])
    for entry in metadata:
        entry['split'] = split_name
        all_metadata.append(entry)

df = pd.DataFrame(all_metadata)

print(f"Total metadata rows loaded: {len(df)}")
display(df.head())
```

```python
df['inquiry'] = df['inquiry'].str.replace('NURC_', '', regex=False)
```

```python
inquiry_list = ["SP_DID_052", "SP_DID_065", "SP_DID_102", "SP_DID_110", "SP_DID_014", "SP_DID_023", "SP_DID_027", "SP_DID_031", "RE_DID_001", "RE_DID_013", "RE_DID_025", "RE_DID_037", "RE_DID_045", "RE_DID_044", "RE_DID_032", "RE_DID_058"]

sub_df = df[df['inquiry'].isin(inquiry_list)]

print(f"Number of rows in sub_df: {len(sub_df)}")
display(sub_df.head())
```

```python
columns_to_keep = ['inquiry', 'segment', 'speaker', 'text']
sub_df = sub_df[columns_to_keep]
print("Sub-dataframe with selected columns:")
display(sub_df.head())
```

```python
output_filename = 'nurc_test_16.csv'
sub_df.to_csv(output_filename, index=False)
print(f"sub_df successfully saved to {output_filename}")
```
